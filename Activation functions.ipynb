```python
5)Experiment with different activation functions (e.g., ReLU, Sigmoid, Tanh) in a simple neural network architecture. Compare their effects on convergence and performance.?

Ans:-
1. Objective

To study how different activation functions — ReLU, Sigmoid, and Tanh — affect the training convergence speed and model performance of a simple neural network.

2. Experimental Setup

Dataset: MNIST Handwritten Digits (or a small dataset like XOR for conceptual demo)

Model Architecture:

Input layer: 784 neurons (28×28 image)

Hidden layer: 64 neurons (activation varied)

Output layer: 10 neurons (Softmax for classification)

Optimizer: Adam

Loss Function: Categorical Cross-Entropy

Epochs: 20

Batch size: 64

3. Code Example (Python - simplified)
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten

# Load dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

# Function to build model with different activations
def build_model(activation):
    model = Sequential([
        Flatten(input_shape=(28, 28)),
        Dense(64, activation=activation),
        Dense(10, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

# Test with different activations
for act in ['sigmoid', 'tanh', 'relu']:
    print(f"\nTraining with {act.upper()} activation:")
    model = build_model(act)
    history = model.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.1, verbose=0)
    loss, acc = model.evaluate(x_test, y_test, verbose=0)
    print(f"{act.upper()} - Test Accuracy: {acc:.4f}")

4. Observations (Example Results)
Activation Function	Convergence Speed	Final Accuracy	Behavior During Training
Sigmoid	Slow	~91%	Gradients vanish for large inputs; learning is slow
Tanh	Moderate	~94%	Faster than sigmoid due to zero-centered outputs
ReLU	Fastest	~97%	Quick convergence, but may have some dead neurons
5. Analysis

Sigmoid:Saturates easily → vanishing gradients.
Training takes longer and may get stuck at local minima.

Tanh:
Performs better than Sigmoid since outputs are zero-centered.
Still suffers from gradient issues for very large or small inputs.

ReLU:
Converges faster and reaches higher accuracy.
Most neurons remain active for positive inputs, allowing efficient gradient flow.
Small risk of “dying ReLU” (neurons output 0 permanently).
```
